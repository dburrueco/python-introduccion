```python
numbers = [3, -2, 4, 1, -6, 2]
```

```python
def mayor_que_cero(number):
    return number > 0
```

```python
filter(mayor_que_cero, numbers)
```

```python
list(filter(mayor_que_cero, numbers))
```

```python
list(filter(lambda number: number < 0, numbers))
```

```python
[number for number in numbers if number > 0]
```
