# map

```python
numbers = list(range(5))
numbers
```

```python
list(map(lambda number: number ** 2, numbers))
```

```python
age = 28
```

```python
"Adulto" if age >= 18 else "No adulto"
```

```python
ages = [5, 25, 67, 14, 33]
```

```python
list(map(lambda age: "Adulto" if age >= 18 else "No adulto", ages))
```

# sorted

```python
sorted(ages)
```

```python
sorted(ages, reverse = True)
```

```python
animales = ["gato", "elefante", "perro", "delfín"]
```

```python
sorted(animales)
```

```python
sorted(animales, key = lambda animal: len(animal))
```

```python
personas = [("Juan", 57), ("Alicia", 34), ("Pedro", 42)]
personas
```

```python
sorted(personas)
```

```python
sorted(personas, key = lambda persona: persona[1])
```
